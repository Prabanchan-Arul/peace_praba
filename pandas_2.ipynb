```python
left=pd.DataFrame({'key':['k0','k1','k2','k3'],
                  'A':['A0','A1','A2','A3'],
                  'B':['B0','B1','B2','B3']})
```

```python
left=pd.DataFrame({'A':['A0','A1','A2'],
                  'B':['B0','B1','B2']},
                  index=['k0','k1','k2'])
```

```python
right=pd.DataFrame({'C':['C0','C1','C2'],
                  'D':['D0','D1','D2']},
                  index=['k0','k2','k3'])
```

```python
left.join(right)
```

```python
right.join(left)
```

```python
left.join(right,how='outer')
```
