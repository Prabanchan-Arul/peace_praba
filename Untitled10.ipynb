```python
from sklearn import preprocessing
label_encoder = preprocessing.LabelEncoder()

columns_to_encode = ['department', 'education', 'recruitment_channel']

for column in columns_to_encode:
  data[column] = label_encoder.fit_transform(data[column])

for column in columns_to_encode:
  print(f"{column}: {data[column].unique()}")
```

```python
data_new=data.drop(['age','gender','employee_id','region'], axis=1)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
X=data_new.drop('is_promoted', axis=1)
y=data_new['is_promoted']
```

```python
# Drop rows with any NaN values before splitting the data
data_new.dropna(inplace=True)

X = data_new.drop('is_promoted', axis=1)
y = data_new['is_promoted']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler
mms=MinMaxScaler(feature_range=(0,1))
X_train=mms.fit_transform(X_train)
X_test=mms.fit_transform(X_test)
X_train=pd.DataFrame(X_train)
X_test=pd.DataFrame(X_test)



X_train.head()
```

```python
X_test.head()
```

```python
from sklearn.metrics import accuracy_score, precision_score, f1_score, confusion_matrix, recall_score
from sklearn.linear_model import LinearRegression, LogisticRegression

# Assuming you have trained a model and made predictions
# Replace LinearRegression with your actual model if it's different
model = LinearRegression()
model1=LogisticRegression()
# Fit the model on the training data
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

model1.fit(X_train, y_train)
y_pred1 = model1.predict(X_test)

# Convert predictions to binary (0 or 1) for classification metrics
y_pred_binary = (y_pred > 0.5).astype(int)
y_pred_binary1 = (y_pred1 > 0.5).astype(int)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred_binary)
precision = precision_score(y_test, y_pred_binary)
f1 = f1_score(y_test, y_pred_binary)
recall = recall_score(y_test, y_pred_binary)
conf_matrix = confusion_matrix(y_test, y_pred_binary)

accuracy1=accuracy_score(y_test, y_pred_binary1)
precision1=precision_score(y_test, y_pred_binary1)
f11=f1_score(y_test, y_pred_binary1)
recall1=recall_score(y_test, y_pred_binary1)
conf_matrix1=confusion_matrix(y_test, y_pred_binary1)


# Print the results
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"Recall: {recall:.4f}")
print("Confusion Matrix:")
print(conf_matrix)

print(f"Accuracy: {accuracy1:.4f}")
print(f"Precision: {precision1:.4f}")
print(f"F1 Score: {f11:.4f}")
print(f"Recall: {recall1:.4f}")
print("Confusion Matrix:")
print(conf_matrix1)
```
